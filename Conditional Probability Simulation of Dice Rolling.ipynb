Suppose you roll a fair six-sided die two times. Let $A$ be the event "the sum of the throws equals 5" and $B$ be the event "at least one of the throws is a $4$".
By solving this problem by hand I got P(A|B)=0.1803. This code creates a simulation to calculate this probability. The more you run it the closer the result gets to the true probability. 
The general idea is to calculate the frequency of an event occuring in a given sample, then dividing the frequency by the size of the 
sample. 

```python
# Simulating 1000 rolls
roll1 = sample(1:6,1000,replace=TRUE)
roll2 = sample(1:6,1000,replace=TRUE)

# Putting the two columns together to create a matrix, this is our sample space
mydata = cbind(roll1,roll2)

# Change from a matrix to a data frame so that we can access elements 
# using the names "roll1" and "roll2"
mydata = as.data.frame(mydata)

# the square brackets can be read as "such that", so we're only accessing the elements 
# that meet the coditions 
# accessing the elements where one of the rolls equals 4, AND the sum of the rolls = 5
# but we're only getting the number of rows (i.e. the frequency) using nrow()
AandB = nrow(mydata[(mydata$roll1==4 | mydata$roll2==4) & (mydata$roll1+mydata$roll2==5),]) 
# accessing the elements where one of the rolls equals 4

# and calculating the number of rows using nrow()
B = nrow(mydata[(mydata$roll1==4 | mydata$roll2==4),])

# calculating the probability of A|B
AgivenB = AandB/B
print(AgivenB)
```
