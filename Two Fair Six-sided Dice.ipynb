Consider rolling two fair six-sided dice. What is the probability that the sum of the results is greater than or equal to 5?

Below is a code I wrote in R to calculate this probability by simulating the experiment (rolling the dice). In this case the simulation is simply generating a random number from 1 to 6. Note that When R sums a vector of booleans (TRUE or FALSE) it is treating TRUE as 1 and FALSE as 0.

```python
numrolls = 10000 #number of rolls
prob = rep(FALSE,numrolls) #a vector containing a FALSE for each roll

for(i in 1:numrolls) 
{
   x1 = sample(1:6, 1, replace=TRUE) #simulating rolling the first die
   x2 = sample(1:6, 1, replace=TRUE) #simulating rolling the 2nd die

   if((x1+x2)>=5) #checking if the sum of the two results is equal to or greater than 5
   {
      prob[i] = TRUE #replacing the FALSE with a TRUE
   }
}

#Summing all the elements in the vector. This will give the total of all the 1s or TRUEs in the vector
sum(prob)/numrolls  #proportion of runs that have a sum equal to or greater than 5
```
