Suppose that $X$ has the binomial distribution with parameters $n=8$ and $p=0.3$. That is, suppose that $X \sim bin (8,0.3)$.
We can calculate probabilities of different values of this random variable by simulating the experiement a number of times, then calculating the proportion of the frequency in which that value appears. We can simulate this distribution using the R function `rbinom(10000,8,0.3)`, then calculate P(X=2).

```python
n = 0
mysample = rbinom(10000,8,0.3)

for (i in 1:length(mysample))
{
    if (mysample[i]==2) #saving the frequency of 2's
    n = n + 1
}
p = n/10000 #calculating the proportion of 2's, which is P(X=2)
print(p) 
```
