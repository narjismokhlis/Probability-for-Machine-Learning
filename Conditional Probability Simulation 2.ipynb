Suppose you have two bags of marbles that are in a box. Bag 1 contains 7 white marbles, 6 black marbles, and 3 gold marbles. Bag 2 contains 4 white marbles, 5 black marbles, and 15 gold marbles. You will reach into the box to pull out a bag. Suppose that, due to the size and shapes of the bags, the probability of grabbing Bag 1 from the box is twice the probability of grabbing Bag 2.

If you close your eyes, grab a bag from the box, and then grab a marble from that bag, what is the probability that it is gold?

calculating the probability by hand: 

first off we can calculate the probabilities of grabbing one of the two bags. Say p is the probability of grabbing bag 2, then the sum of probabilities gives the following equation: 2p + p = 1, solving for p we get p=1/3, so P(B1)=2/3 and P(B2)=1/3.

Now we can calculate the probability of grabbing a gold marble from each one of the two bags, which are the conditional probabilities:

- Grabbing a gold marble knowing bag 1 was grabbed: P(G|B1) = (3!/1!*2!)*1/16 = 3/16
- Grabbing a gold marble knowing bag 2 was grabbed: P(G|B2) = (15!/1!*14!)*1/24 = 15/24

Now we can use the law of total probability to calculate P(G), since B2 and B1 are complements of each other: 

P(G) = P(G|B1) * P(B1) + P(G|B2) *P(B2) = 0.3332

I wrote this code to simulate the experiment of randomly grabbing a marble from a bag, and calculate the probability of the marble being gold. This code gives a closer result to the analytical value the more you run it. 

```python
PB1 = 2/3
PB2 = 1/3

numreps = 1000
GB1 = rep(FALSE,numreps)
GB2 = rep(FALSE,numreps)

for (i in 1:numreps)
{
   SB1 = sample(c("white","black","gold"),1,prob=c(7/16,6/16,3/16))
   if (SB1 == 'gold')
       {
       GB1[i]=TRUE
       } 
    
   SB2 = sample(c("white","black","gold"),1,prob=c(4/24,5/24,15/24))
   if (SB2 == 'gold')
       {
       GB2[i]=TRUE
       }
}
PGB1 = sum(GB1)/numreps
PGB2 = sum(GB2)/numreps

PG = (PGB1 * PB1) + (PGB2 * PB2)
```
